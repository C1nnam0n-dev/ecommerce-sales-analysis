```python
df = pd.read_csv("../data/ecommerce_sales_analytics_5000.csv")
df.head()
df.shape
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df.info()
```

```python
df.head()
```

```python
df["product_category"].value_counts()
```

```python
df.groupby("product_category")["revenue"].sum().sort_values(ascending=False)
```

```python
revenue_by_category = df.groupby("product_category")["revenue"].sum().sort_values(ascending=False)
revenue_by_category.plot(kind="bar")
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../Images/revenue_by_product_category.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
df.groupby("region")["revenue"].sum().sort_values(ascending=False)
```

```python
region_revenue = df.groupby("region")["revenue"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
region_revenue.plot(kind="bar")
plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../Images/revenue_by_region.png")
plt.show()
```

```python
monthly_revenue = df.groupby(df["order_date"].dt.to_period("M"))["revenue"].sum()
monthly_revenue
```

```python
plt.figure(figsize=(10, 5))
monthly_revenue.plot(kind="line")
plt.title("Monthly Revenue Over Time")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../Images/revenue_over_time.png")
plt.show()
```

```python
payment_revenue = df.groupby("payment_method")["revenue"].sum().sort_values(ascending=False)
payment_revenue
```

```python
plt.figure(figsize=(8, 5))
payment_revenue.plot(kind="bar")
plt.title("Revenue by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../Images/revenue_by_payment_method.png")
plt.show()
```

```python
rating_by_category = df.groupby("product_category")["customer_rating"].mean().sort_values(ascending=False)
rating_by_category
```

```python
rating_by_category.plot(kind="bar", figsize=(8,5))
plt.title("Average Customer Rating by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Average Customer Rating")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../Images/average_rating_by_category.png")
plt.show()
```

```python
delivery_by_category = df.groupby("product_category")["delivery_days"].mean().sort_values()
delivery_by_category
```

```python
delivery_by_category.plot(kind="bar", figsize=(8,5))
plt.title("Average Delivery Time by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Delivery Time")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../Images/average_delivery_by_category.png")
plt.show()
```

```python
delivery_by_region = df.groupby("region")["delivery_days"].mean().sort_values()
delivery_by_region
```

```python
delivery_by_region.plot(kind="bar", figsize=(8, 5))
plt.title("Average Delivery Time by Region")
plt.xlabel("Region")
plt.ylabel("Average Delivery Time (Days)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../Images/average_delivery_by_region.png")
plt.show()
```

```python
region_category_revenue = df.groupby(["region", "product_category"])["revenue"].sum().unstack()
region_category_revenue
```

```python
region_category_revenue.plot(kind="bar", figsize=(10,6))
plt.title("Revenue by Region and Product Category")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../Images/revenue_by_region_category.png")
plt.show()
```
